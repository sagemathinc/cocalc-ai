# Move and reproduce a notebook analysis

Keep `measurements.csv`, `analyze.py`, and this notebook in the same directory. This example needs a local Python 3 kernel, with no additional analysis packages. Run every cell from a fresh kernel; saved output alone is not a rerun.


```python
from pathlib import Path
import sys

print("Python:", sys.version.split()[0])
print("Working directory:", Path.cwd())
assert Path("measurements.csv").is_file(), "Upload measurements.csv beside the notebook"
assert Path("analyze.py").is_file(), "Upload analyze.py beside the notebook"
```

## Analyze and check

The synthetic values are 2, 4, 6, and 8. The expected mean is 5.0. The input hash lets another researcher check that their input bytes match.


```python
import json
from analyze import analyze

result = analyze("measurements.csv")
assert result["count"] == 4
assert result["mean"] == 5.0
Path("notebook-result.json").write_text(json.dumps(result, indent=2) + "\n")
print(f"count={result['count']} mean={result['mean']:.1f}")
```

## Check saved results

Reload the JSON from disk instead of relying on a Python variable. Then restart the kernel and run every cell again. Compare the saved input hash and results with the terminal workflow.


```python
saved = json.loads(Path("notebook-result.json").read_text())
assert saved == result
print("Saved result matches the computation")
```
